```python
!wget -O survey-data.csv https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv

# Step 2: Import necessary libraries and load the dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
df = pd.read_csv("survey-data.csv")

# Display the first few rows to understand the structure of the data
df.head()
```

```python
#Top 10 current languages
```

```python
current_langs = df['LanguageHaveWorkedWith'].dropna().str.split(';').explode()
top_10_current = current_langs.value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_10_current.index, y=top_10_current.values, color='#b65100')
plt.title("Top 10 Programming Languages")
plt.xlabel("Programming Languages")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45, ha='right', fontsize=11)
plt.yticks(fontsize=11)
plt.tight_layout()
plt.show()
```

```python
#Top 10 future languages
```

```python
future_langs = df['LanguageWantToWorkWith'].dropna().str.split(';').explode()

top_10_future = future_langs.value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_10_future.index, y=top_10_future.values, color='#b65100')
plt.title("Top 10 Anticipated Programming Languages for Next Year")
plt.xlabel("Programming Languages")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45, ha='right', fontsize=11)
plt.yticks(fontsize=11)
plt.tight_layout()
plt.show()
```

```python
#Top 10 Databases in use
```
