### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
age_order = ['Under 18 years old', '18-24 years old', '25-34 years old', '35-44 years old', '45-54 years old', '55-64 years old', '65 years or older']
med = df.groupby('Age')[['JobSatPoints_6', 'JobSatPoints_7']].median().reindex(age_order)
med.plot(kind='bar', stacked=True, figsize=(10, 6), color=['#6a2c00', '#ffd54f'])
plt.legend(loc='upper left')
plt.title('Median JobSatPoints_6 and JobSatPoints_7 by Age Group')
plt.xlabel('Age group')
plt.ylabel('Median score (stacked)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


```python
top_emp = df['Employment'].value_counts().head(6).index
med = df[df['Employment'].isin(top_emp)].groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].median()
med.plot(kind='bar', stacked=True, figsize=(11, 6), color=['#6a2c00', '#ffd54f'])
plt.legend(loc='upper right')
plt.title('Median JobSatPoints_6 and JobSatPoints_7 by Employment Status')
plt.xlabel('Employment')
plt.ylabel('Median score (stacked)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
age_map = {'Under 18 years old': 16, '18-24 years old': 21, '25-34 years old': 30, '35-44 years old': 40, '45-54 years old': 50, '55-64 years old': 60, '65 years or older': 70, 'Prefer not to say': None}
df['Age_numeric'] = df['Age'].map(age_map)
grp = df[(df['Age_numeric'] >= 30) & (df['Age_numeric'] <= 35)]
comp_k = grp['ConvertedCompYearly'].median() / 1000
jobsat = grp['JobSat'].median()
plt.figure(figsize=(5, 6))
plt.bar('Ages 30-35', comp_k, label='Median comp ($000s)', color='#6a2c00')
plt.bar('Ages 30-35', jobsat, bottom=comp_k, label='Median JobSat', color='#ffd54f')
plt.title('Median Compensation and Job Satisfaction (Ages 30-35)')
plt.ylabel('Stacked value')
plt.legend(loc='lower right')
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
age_order = ['Under 18 years old', '18-24 years old', '25-34 years old', '35-44 years old', '45-54 years old', '55-64 years old', '65 years or older']
summary = df.groupby('Age').agg(comp_k=('ConvertedCompYearly', lambda x: x.median() / 1000), jobsat=('JobSat', 'median')).reindex(age_order)
summary.plot(kind='bar', stacked=True, figsize=(10, 6), color=['#6a2c00', '#ffd54f'])
plt.legend(loc='upper left')
plt.title('Median Compensation ($000s) and Job Satisfaction by Age Group')
plt.xlabel('Age group')
plt.ylabel('Stacked value')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
age_order = ['18-24 years old', '25-34 years old', '35-44 years old', '45-54 years old', '55-64 years old']
top_dbs = df['DatabaseWantToWorkWith'].dropna().str.split(';').explode().str.strip().value_counts().head(4).index
exploded = df[['Age', 'DatabaseWantToWorkWith']].dropna().copy()
exploded['DatabaseWantToWorkWith'] = exploded['DatabaseWantToWorkWith'].str.split(';')
exploded = exploded.explode('DatabaseWantToWorkWith')
exploded['DatabaseWantToWorkWith'] = exploded['DatabaseWantToWorkWith'].str.strip()
exploded = exploded[exploded['DatabaseWantToWorkWith'].isin(top_dbs)]
pivot = exploded.pivot_table(index='Age', columns='DatabaseWantToWorkWith', aggfunc='size').reindex(age_order)
pivot.plot(kind='bar', stacked=True, figsize=(11, 6), color=['#fff9c4', '#ffb300', '#b65100', '#211000'], edgecolor='white', linewidth=0.5)
plt.legend(loc='upper right')
plt.title('Preferred Databases by Age Group (top 4)')
plt.xlabel('Age group')
plt.ylabel('Number of respondents')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
top_emp = df['Employment'].value_counts().head(5).index
sub = df[df['Employment'].isin(top_emp)].dropna(subset=['JobSat'])
pivot = sub.pivot_table(index='JobSat', columns='Employment', aggfunc='size', fill_value=0)
pivot.plot(kind='bar', stacked=True, figsize=(11, 6), color=['#ffb74d', '#f57c00', '#b65100', '#6d2d00', '#211000'], edgecolor='white', linewidth=0.5)
plt.legend(loc='upper left', fontsize=11)
plt.title('Employment Type by Job Satisfaction Level')
plt.xlabel('JobSat')
plt.ylabel('Number of respondents')
plt.tight_layout()
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
age_order = ['18-24 years old', '25-34 years old', '35-44 years old', '45-54 years old', '55-64 years old']
top_langs = df['LanguageAdmired'].dropna().str.split(';').explode().str.strip().value_counts().head(4).index
exploded = df[['Age', 'LanguageAdmired']].dropna().copy()
exploded['LanguageAdmired'] = exploded['LanguageAdmired'].str.split(';')
exploded = exploded.explode('LanguageAdmired')
exploded['LanguageAdmired'] = exploded['LanguageAdmired'].str.strip()
exploded = exploded[exploded['LanguageAdmired'].isin(top_langs)]
pivot = exploded.pivot_table(index='Age', columns='LanguageAdmired', aggfunc='size').reindex(age_order)
pivot.plot(kind='bar', stacked=True, figsize=(11, 6), color=['#ffb74d', '#f57c00', '#b65100', '#211000'], edgecolor='white', linewidth=0.5)
plt.legend(loc='upper right')
plt.title('Admired Programming Languages by Age Group (Top 4)')
plt.xlabel('Age group')
plt.ylabel('Number of respondents')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
top_emp = df['Employment'].value_counts().head(4).index
top_plat = df['PlatformAdmired'].dropna().str.split(';').explode().str.strip().value_counts().head(4).index
exploded = df[df['Employment'].isin(top_emp)][['Employment', 'PlatformAdmired']].dropna().copy()
exploded['PlatformAdmired'] = exploded['PlatformAdmired'].str.split(';')
exploded = exploded.explode('PlatformAdmired')
exploded['PlatformAdmired'] = exploded['PlatformAdmired'].str.strip()
exploded = exploded[exploded['PlatformAdmired'].isin(top_plat)]
pivot = exploded.pivot_table(index='Employment', columns='PlatformAdmired', aggfunc='size')
pivot.plot(kind='bar', stacked=True, figsize=(11, 6), color=['#ffb74d', '#f57c00', '#b65100', '#211000'], edgecolor='white', linewidth=0.5)
plt.legend(loc='upper right')
plt.title('Admired Platforms by Employment Type (top 4)')
plt.xlabel('Employment')
plt.ylabel('Number of respondents')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```
