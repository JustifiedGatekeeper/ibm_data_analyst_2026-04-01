## Tasks


<h2>1. Load the Dataset</h2>
<p>

The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
print(df.info())
print("\nSummary statistics:")
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
missing_values = df.isnull().sum().sort_values(ascending=False).head(10)
print(missing_values)
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
plt.figure(figsize=(14, 7))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap='viridis') #I wanted to see if I can use a different colour scheme
plt.title('Missing Values Heatmap')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
missing_employment_count = df['Employment'].isnull().sum()
print(f"Total missing rows for Employment: {missing_employment_count}")
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
majority_employment_type = df['Employment'].mode()[0]
print(f"The most frequent employment status: {majority_employment_type}")
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
df['Employment'] = df['Employment'].fillna(majority_employment_type)

print(f"Empty rows left in Employment: {df['Employment'].isnull().sum()}")
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
employment_count = df['Employment'].value_counts()

employment_count.plot(kind='bar', figsize=(10, 5), color='orange')

plt.title('Distribution of Employment after imputation')
plt.xlabel('Employment')
plt.ylabel('Total Count')
plt.xticks(rotation=90, ha='right')
plt.tight_layout()
plt.show()
```

```python
employment_count = df['Employment'].value_counts()

employment_count.plot(kind='pie', figsize=(8, 8), autopct='%1.1f%%', startangle=90)

plt.title('Distribution of Employment after imputation', fontsize=14, fontweight='bold')
plt.ylabel('')

plt.tight_layout()
plt.show()
```

```python
employment_count = df['Employment'].value_counts()

total_responses = employment_count.sum()
under_one_percent = total_responses * 0.01

big_slices = employment_count[employment_count >= under_one_percent]
tiny_slices = employment_count[employment_count < under_one_percent]

final_counts = big_slices.copy()
final_counts['Other'] = tiny_slices.sum()

final_counts.plot(kind='pie', figsize=(8, 8), autopct='%1.1f%%', startangle=90)

plt.title('Distribution of Employment after imputation', fontsize=14, fontweight='bold')
plt.ylabel('')

plt.tight_layout()
plt.show()
```
